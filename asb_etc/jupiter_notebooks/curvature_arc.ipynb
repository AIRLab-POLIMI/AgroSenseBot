```python
from numpy import *
from matplotlib import pyplot as plt
plt.rcParams['figure.figsize'] = [7, 7]
from matplotlib.patches import Arc

# x_g, y_g, t: goal pose (t is theta)
# x_p, y_p: intersection point between goal pose y-axis and robot pose y-axis
# x_a, y_a: angle goal pose in ASB RPP, new pose translated on the goal pose's y-axis such that 
#     the final angle on the lookahead arc is the same as the goal angle
# c: curvature that leads the robot to the x_a, y_a pose

def plot(c, s, t):
    x_p = 0
    y_p = 1 / c if abs(c) > 0 else inf

    print(f"c = {c:.4f} t = {t:.4f}")
    print(f"x_p = {x_p:.4f} y_p = {y_p:.4f}")

    fig, ax = plt.subplots()
    
    plt.scatter(x_p, y_p, color='lime')

    if abs(y_p) < inf:
        r = abs(y_p)
        t_1 = pi/2 if y_p < 0 else -pi/2
        t_2 = t_1 + t
        
        print(f"t_1 = {t_1:.4f} t_2 = {t_2:.4f}")
        
        arc_angles = linspace(min(t_1, t_2), max(t_1, t_2), 10)
        arc_xs = x_p + r * cos(arc_angles)
        arc_ys = y_p + r * sin(arc_angles)
        # print(arc_angles)
        plt.plot(arc_xs, arc_ys, color='red', lw = 3)
    
    # plt.xlim(-1.5, 1.5)
    # plt.ylim(-1.5, 1.5)
    plt.grid()
    # ax.axhline(y=0, color='k')
    # ax.axvline(x=0, color='k')
    ax.set_aspect('equal', adjustable='box')
    # plt.figure().set_figwidth(15)
    plt.show()

# plot(c=.1, s=1, t=0.1 * pi)
# plot(c=-.1, s=1, t=0.1 * pi)
# plot(c=.1, s=1, t=-0.1 * pi)
# plot(c=-.1, s=1, t=-0.1 * pi)
plot(c=-0.241, s=1, t=-0.112)

# curvature: -0.241   final_angle:  -0.112   y_p:  -4.144   theta_1:  1.000   theta_2:  0.888
```

```python

def plot_l(x, y):
    
    l = sqrt(x**2 + y**2)
    c = 2 * y / l**2
    w = sign(x)
    
    print(f"x = {x:.4f} y = {y:.4f}")
    print(f"c = {c:.4f} l = {l:.4f}")
    fig, ax = plt.subplots()
    
    d = linspace(0, l, 100)
    arc_xs = w * sqrt(d**2 - c**2 * d**4 / 4)
    arc_ys = c * d**2 / 2
    plt.plot(arc_xs, arc_ys, color='red', lw = 3)

    plt.grid()
    ax.set_aspect('equal', adjustable='box')
    plt.show()

plot_l(1, 0.5)
plot_l(1, -0.5)
plot_l(-1, 0.5)
plot_l(-1, -0.5)


plot_l(1, 1)
plot_l(1, -1)
plot_l(-1, 1)
plot_l(-1, -1)
```
