```python
import pandas as pd
from numpy import *
%matplotlib ipympl
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = [7, 7]
import os
```

```python
df = pd.read_csv(os.path.expanduser("/home/enrico/tmp/p_2025-05-13/p_2025-05-13_15:42:22.252959.csv"))
df.t = df.t*1E-6  # convert t (in micro seconds) to seconds
df["p_bar"] = df.p*10  # add pressure in bars (p is in MPa)
df.t_msg = pd.to_datetime(df.t_msg)
df.set_index("t_msg", inplace=True)
print(df)

d=df
```

```python
plt.cla()
plt.ylabel("v [V]")

plt.plot(d.index, d.v, label="orig", color="lightgrey")

fast_window = 50
slow_window = 200

# plt.plot(d.index, d.v.rolling(window=f"{fast_window}ms", center=True).mean(), label=f"mean {fast_window} ms")
plt.plot(d.index, d.v.rolling(window=f"{fast_window}ms", center=True).median(), label=f"median {fast_window} ms")

plt.plot(d.index, d.v.rolling(window=f"{slow_window}ms", center=True).mean(), label=f"mean {slow_window} ms")
plt.plot(d.index, d.v.rolling(window=f"{slow_window}ms", center=True).median(), label=f"median {slow_window} ms")

plt.grid()
plt.legend()
plt.show()
```

```python
plt.cla()
plt.ylabel("p [bar]")

plt.plot(d.index, d.p_bar, label="orig", color="lightgrey")

fast_window = 50
slow_window = 200

# plt.plot(d.index, d.p_bar.rolling(window=f"{fast_window}ms", center=True).mean(), label=f"mean {fast_window} ms")
plt.plot(d.index, d.p_bar.rolling(window=f"{fast_window}ms", center=True).median(), label=f"median {fast_window} ms")

# plt.plot(d.index, d.p_bar.rolling(window=f"{slow_window}ms", center=True).mean(), label=f"mean {slow_window} ms")
plt.plot(d.index, d.p_bar.rolling(window=f"{slow_window}ms", center=True).median(), label=f"median {slow_window} ms")

plt.grid()
plt.legend()
plt.show()
```
